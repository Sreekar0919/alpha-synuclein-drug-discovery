```python
import requests
from Bio import SeqIO
import io

def load_alpha_synuclein():
    url = "https://www.uniprot.org/uniprot/P37840.fasta"
    response = requests.get(url)
    if response.status_code == 200:
        fasta_io = io.StringIO(response.text)
        for record in SeqIO.parse(fasta_io, "fasta"):
            return str(record.seq)
    else:
        raise Exception("Failed to fetch alpha-synuclein sequence from UniProt")

# Load sequence
alpha_synuclein_sequence = load_alpha_synuclein()
print(alpha_synuclein_sequence)
```
