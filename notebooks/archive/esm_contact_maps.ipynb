```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np
from Bio.SeqUtils.ProtParam import ProteinAnalysis

# Alpha-synuclein sequence
original_sequence = "MDVFMKGLSKAKEGVVAAAEKTKQGVAEAAGKTKEGVLYVGSKTKEGVVHGVATVAEKTKEQVTNVGGAVVTGVTAVAQKTVEGAGSIAAATGFVKKDQLGKNEEGAPQEGILEDMPVDPDNEAYEMPSEEGYQDYEPEA"

# Function to convert sequence to one-hot encoding
def sequence_to_onehot(sequence):
    aa_dict = {aa: i for i, aa in enumerate('ACDEFGHIKLMNPQRSTVWY')}
    onehot = torch.zeros(len(sequence), 20)
    for i, aa in enumerate(sequence):
        if aa in aa_dict:
            onehot[i, aa_dict[aa]] = 1
    return onehot

# Simple CNN model for sequence analysis
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        self.conv1 = nn.Conv1d(20, 64, kernel_size=3, padding=1)
        self.conv2 = nn.Conv1d(64, 128, kernel_size=3, padding=1)
        self.fc = nn.Linear(128, 20)

    def forward(self, x):
        x = torch.relu(self.conv1(x))
        x = torch.relu(self.conv2(x))
        x = x.mean(dim=2)
        return self.fc(x)

# Function to analyze sequence
def analyze_sequence(sequence):
    onehot = sequence_to_onehot(sequence)
    model = SimpleCNN()
    output = model(onehot.unsqueeze(0).transpose(1, 2))
    return output.detach().numpy()

# Function to calculate protein properties
def calculate_properties(sequence):
    analysis = ProteinAnalysis(sequence)
    return {
        "Molecular Weight": analysis.molecular_weight(),
        "Aromaticity": analysis.aromaticity(),
        "Instability Index": analysis.instability_index(),
        "Isoelectric Point": analysis.isoelectric_point()
    }

# Analyze original sequence
print("Original Sequence Analysis:")
original_analysis = analyze_sequence(original_sequence)
original_properties = calculate_properties(original_sequence)

plt.figure(figsize=(10, 5))
plt.imshow(original_analysis, cmap='viridis')
plt.title("Sequence Analysis of Original Alpha-Synuclein")
plt.colorbar()
plt.show()

for key, value in original_properties.items():
    print(f"{key}: {value}")

# Example mutation (replace this with your own mutation strategy)
mutated_sequence = original_sequence[:120] + "AAAAAA" + original_sequence[126:]

# Analyze mutated sequence
print("\nMutated Sequence Analysis:")
mutated_analysis = analyze_sequence(mutated_sequence)
mutated_properties = calculate_properties(mutated_sequence)

plt.figure(figsize=(10, 5))
plt.imshow(mutated_analysis, cmap='viridis')
plt.title("Sequence Analysis of Mutated Alpha-Synuclein")
plt.colorbar()
plt.show()

for key, value in mutated_properties.items():
    print(f"{key}: {value}")

# Compare properties
print("\nProperty Changes:")
for key in original_properties:
    change = mutated_properties[key] - original_properties[key]
    print(f"{key} change: {change}")
```

```python
import torch
import sys
import matplotlib.pyplot as plt
import numpy as np
from Bio.SeqUtils.ProtParam import ProteinAnalysis

# Add the ESM library to the Python path
sys.path.append('/path/to/esm')  # Replace with the actual path to the ESM directory

from esm import FastaBatchedDataset, pretrained

# Load the ESM model
model, alphabet = pretrained.load_model_and_alphabet("esm3_t66_650M_UR50S")
model.eval()  # Set the model to evaluation mode

# Alpha-synuclein sequence
original_sequence = "MDVFMKGLSKAKEGVVAAAEKTKQGVAEAAGKTKEGVLYVGSKTKEGVVHGVATVAEKTKEQVTNVGGAVVTGVTAVAQKTVEGAGSIAAATGFVKKDQLGKNEEGAPQEGILEDMPVDPDNEAYEMPSEEGYQDYEPEA"

def analyze_sequence(sequence, title):
    # Prepare data
    dataset = FastaBatchedDataset.from_sequences([(title, sequence)])
    batches = dataset.get_batch_indices(1, len(sequence))
    batch_labels, batch_strs, batch_tokens = next(batches)
    
    # Extract embeddings
    with torch.no_grad():
        results = model(batch_tokens, repr_layers=[66])
    
    # Get per-residue representations (embeddings)
    token_representations = results["representations"][66]
    
    # Visualize embeddings
    plt.figure(figsize=(10, 5))
    plt.imshow(token_representations[0].T.cpu(), aspect='auto', interpolation='nearest')
    plt.title(f"ESM Embeddings - {title}")
    plt.colorbar()
    plt.show()
    
    return token_representations

# Function to calculate protein properties
def calculate_properties(sequence):
    analysis = ProteinAnalysis(sequence)
    return {
        "Molecular Weight": analysis.molecular_weight(),
        "Aromaticity": analysis.aromaticity(),
        "Instability Index": analysis.instability_index(),
        "Isoelectric Point": analysis.isoelectric_point()
    }

# Analyze original sequence
print("Original Sequence Analysis:")
original_embeddings = analyze_sequence(original_sequence, "Original Alpha-Synuclein")
original_properties = calculate_properties(original_sequence)
for key, value in original_properties.items():
    print(f"{key}: {value}")

# Example mutation (replace this with your own mutation strategy)
mutated_sequence = original_sequence[:120] + "AAAAAA" + original_sequence[126:]

# Analyze mutated sequence
print("\nMutated Sequence Analysis:")
mutated_embeddings = analyze_sequence(mutated_sequence, "Mutated Alpha-Synuclein")
mutated_properties = calculate_properties(mutated_sequence)
for key, value in mutated_properties.items():
    print(f"{key}: {value}")

# Compare properties
print("\nProperty Changes:")
for key in original_properties:
    change = mutated_properties[key] - original_properties[key]
    print(f"{key} change: {change}")

# Compare embeddings
embedding_diff = torch.mean(torch.abs(original_embeddings - mutated_embeddings))
print(f"\nMean absolute difference in embeddings: {embedding_diff.item()}")
```
